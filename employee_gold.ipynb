```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df_emp_silver=spark.read.table('emp_silver')
```

```python
df_emp_silver.display()
```

```python
df_emp_gold=df_emp_silver.groupBy("JOB").agg(sum("SAL").alias('Tot_sal'))
df_emp_gold.display()
```

```python
df_emp_gold.write.saveAsTable('emp_gold')
```
